```python
# EXPERIMENT 26
# Insert Watermark to Image using OpenCV

import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
filename = next(iter(uploaded))

# Read image
img = cv2.imread(filename)

# Copy original image
watermarked = img.copy()

# Watermark text
text = "M.VIGNESH"

# Position
position = (30, 50)

# Font
font = cv2.FONT_HERSHEY_SIMPLEX

# Font size
font_scale = 1

# Thickness
thickness = 2

# Watermark color (white)
color = (255, 255, 255)

# Add text to image
overlay = watermarked.copy()

cv2.putText(
    overlay,
    text,
    position,
    font,
    font_scale,
    color,
    thickness,
    cv2.LINE_AA
)

# Transparency
alpha = 0.5

# Blend overlay and original
watermarked = cv2.addWeighted(
    overlay,
    alpha,
    watermarked,
    1 - alpha,
    0
)

# Convert BGR to RGB
original_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
watermarked_rgb = cv2.cvtColor(watermarked, cv2.COLOR_BGR2RGB)

# Display
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(original_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(watermarked_rgb)
plt.title("Watermarked Image")
plt.axis("off")

plt.show()
```
