```python
# EXPERIMENT 24
# Sharpening of Image using High-Boost Mask

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
filename = next(iter(uploaded))

# Read image
img = cv2.imread(filename)

# Convert to RGB for displaying
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Blur the image
blurred = cv2.GaussianBlur(img, (5, 5), 0)

# High-boost filtering
# High-boost formula: g = A*f - blurred
A = 2.0

high_boost = cv2.addWeighted(img, A, blurred, -(A - 1), 0)

# Convert result to RGB
high_boost_rgb = cv2.cvtColor(high_boost, cv2.COLOR_BGR2RGB)

# Display
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(high_boost_rgb)
plt.title("High-Boost Sharpened Image")
plt.axis("off")

plt.show()
```
