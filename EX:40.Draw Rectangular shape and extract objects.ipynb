```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

filename = list(uploaded.keys())[0]

# Read image
img = cv2.imread(filename)

# Display image
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 7))
plt.imshow(img_rgb)
plt.axis("on")
plt.title("Original Image")
plt.show()

# Get image dimensions
height, width = img.shape[:2]

print("Image Width:", width)
print("Image Height:", height)
```

```python
# Rectangle coordinates
x = 100
y = 100
w = 300
h = 300

# Draw rectangle
rect_img = img.copy()

cv2.rectangle(
    rect_img,
    (x, y),
    (x+w, y+h),
    (0, 255, 0),
    3
)

# Display rectangle
rect_rgb = cv2.cvtColor(rect_img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 7))
plt.imshow(rect_rgb)
plt.axis("off")
plt.title("Rectangular Selection")
plt.show()
```

```python
# Extract selected object
extracted_object = img[y:y+h, x:x+w]

# Convert to RGB
extracted_rgb = cv2.cvtColor(
    extracted_object,
    cv2.COLOR_BGR2RGB
)

# Display extracted object
plt.figure(figsize=(6, 6))
plt.imshow(extracted_rgb)
plt.axis("off")
plt.title("Extracted Object")
plt.show()

# Save extracted object
cv2.imwrite(
    "extracted_object.jpg",
    extracted_object
)

print("Object extracted successfully!")
```
