```python
# EXPERIMENT 13: Perspective Transformation on Video

import cv2
import numpy as np
from google.colab import files
from IPython.display import HTML
from base64 import b64encode

# Step 1: Upload video
print("Upload your video file:")
uploaded = files.upload()

video_name = list(uploaded.keys())[0]

# Step 2: Open video
cap = cv2.VideoCapture(video_name)

# Get video properties
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

# If FPS is not detected
if fps == 0:
    fps = 30

# Step 3: Define 4 source points
src_points = np.float32([
    [0, 0],
    [width - 1, 0],
    [width - 1, height - 1],
    [0, height - 1]
])

# Step 4: Define 4 destination points
dst_points = np.float32([
    [80, 50],
    [width - 80, 30],
    [width - 40, height - 50],
    [40, height - 30]
])

# Step 5: Calculate perspective transformation matrix
M = cv2.getPerspectiveTransform(src_points, dst_points)

# Step 6: Create output video
output_file = "perspective_transformed_video.mp4"

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

out = cv2.VideoWriter(
    output_file,
    fourcc,
    fps,
    (width, height)
)

# Step 7: Process every video frame
while True:

    ret, frame = cap.read()

    if not ret:
        break

    # Apply perspective transformation
    transformed_frame = cv2.warpPerspective(
        frame,
        M,
        (width, height)
    )

    # Save transformed frame
    out.write(transformed_frame)

# Step 8: Release video
cap.release()
out.release()

print("✅ Perspective Transformation Completed!")
print("Output:", output_file)

# Step 9: Display output video in Colab
video_data = open(output_file, 'rb').read()
video_base64 = b64encode(video_data).decode()

HTML(f'''
<video width="700" controls>
    <source src="data:video/mp4;base64,{video_base64}" type="video/mp4">
</video>
''')
```
