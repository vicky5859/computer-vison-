```python
# ============================================================
# EXERCISE 3: NOISE CHARACTERIZATION AND FILTER SELECTION
# 9x9 Linear Gaussian Filter vs 9x9 Non-Linear Median Filter
# ============================================================

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# ------------------------------------------------------------
# STEP 1: Upload Image
# ------------------------------------------------------------

print("Upload an image:")
uploaded = files.upload()

filename = list(uploaded.keys())[0]

# Read image
img = cv2.imread(filename)

# Check image
if img is None:
    print("Error: Image could not be loaded.")
else:

    # Convert BGR to RGB
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    print("Image loaded successfully!")
    print("Image size:", img.shape)

    # --------------------------------------------------------
    # STEP 2: Display Original Image
    # --------------------------------------------------------

    plt.figure(figsize=(7, 5))
    plt.imshow(img)
    plt.title("Original Image")
    plt.axis("off")
    plt.show()

    # ========================================================
    # CASE A: WHITE GAUSSIAN NOISE
    # ========================================================

    print("\nCASE A: White Gaussian Noise")

    # Convert to float
    img_float = img.astype(np.float32)

    # Generate Gaussian noise
    mean = 0
    sigma = 25

    gaussian_noise = np.random.normal(
        mean,
        sigma,
        img.shape
    )

    # Add noise
    noisy_gaussian = img_float + gaussian_noise

    # Keep pixel values between 0 and 255
    noisy_gaussian = np.clip(
        noisy_gaussian,
        0,
        255
    ).astype(np.uint8)

    # --------------------------------------------------------
    # STEP 3: Apply 9x9 Linear Gaussian Filter
    # --------------------------------------------------------

    gaussian_filtered = cv2.GaussianBlur(
        noisy_gaussian,
        (9, 9),
        0
    )

    # --------------------------------------------------------
    # STEP 4: Apply 9x9 Non-Linear Median Filter
    # --------------------------------------------------------

    median_on_gaussian = cv2.medianBlur(
        noisy_gaussian,
        9
    )

    # --------------------------------------------------------
    # Display Case A Results
    # --------------------------------------------------------

    plt.figure(figsize=(16, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(noisy_gaussian)
    plt.title("Case A - Gaussian Noise")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(gaussian_filtered)
    plt.title("9x9 Linear Gaussian Filter")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(median_on_gaussian)
    plt.title("9x9 Non-Linear Median Filter")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # ========================================================
    # CASE B: SALT AND PEPPER / IMPULSE NOISE
    # ========================================================

    print("\nCASE B: Salt and Pepper Noise")

    # Copy original
    salt_pepper = img.copy()

    # Noise percentage
    noise_amount = 0.08

    # Total number of pixels
    total_pixels = img.shape[0] * img.shape[1]

    # Number of noisy pixels
    num_noisy_pixels = int(
        noise_amount * total_pixels
    )

    # --------------------------------------------------------
    # Add Salt Noise (White Pixels)
    # --------------------------------------------------------

    for i in range(num_noisy_pixels // 2):

        y = np.random.randint(
            0,
            img.shape[0]
        )

        x = np.random.randint(
            0,
            img.shape[1]
        )

        salt_pepper[y, x] = [255, 255, 255]

    # --------------------------------------------------------
    # Add Pepper Noise (Black Pixels)
    # --------------------------------------------------------

    for i in range(num_noisy_pixels // 2):

        y = np.random.randint(
            0,
            img.shape[0]
        )

        x = np.random.randint(
            0,
            img.shape[1]
        )

        salt_pepper[y, x] = [0, 0, 0]

    # --------------------------------------------------------
    # STEP 5: Gaussian Filter on Salt & Pepper Noise
    # --------------------------------------------------------

    sp_gaussian_filtered = cv2.GaussianBlur(
        salt_pepper,
        (9, 9),
        0
    )

    # --------------------------------------------------------
    # STEP 6: Median Filter on Salt & Pepper Noise
    # --------------------------------------------------------

    sp_median_filtered = cv2.medianBlur(
        salt_pepper,
        9
    )

    # --------------------------------------------------------
    # Display Case B Results
    # --------------------------------------------------------

    plt.figure(figsize=(16, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(salt_pepper)
    plt.title("Case B - Salt & Pepper Noise")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(sp_gaussian_filtered)
    plt.title("9x9 Linear Gaussian Filter")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(sp_median_filtered)
    plt.title("9x9 Non-Linear Median Filter")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # ========================================================
    # FINAL COMPARISON
    # ========================================================

    plt.figure(figsize=(18, 10))

    # Original
    plt.subplot(2, 4, 1)
    plt.imshow(img)
    plt.title("Original Image")
    plt.axis("off")

    # Gaussian noise
    plt.subplot(2, 4, 2)
    plt.imshow(noisy_gaussian)
    plt.title("Gaussian Noise")
    plt.axis("off")

    # Gaussian filtered
    plt.subplot(2, 4, 3)
    plt.imshow(gaussian_filtered)
    plt.title("Gaussian Filter 9x9")
    plt.axis("off")

    # Median filtered Gaussian
    plt.subplot(2, 4, 4)
    plt.imshow(median_on_gaussian)
    plt.title("Median Filter 9x9")
    plt.axis("off")

    # Salt & pepper
    plt.subplot(2, 4, 5)
    plt.imshow(salt_pepper)
    plt.title("Salt & Pepper Noise")
    plt.axis("off")

    # Gaussian filter on S&P
    plt.subplot(2, 4, 6)
    plt.imshow(sp_gaussian_filtered)
    plt.title("Gaussian Filter 9x9")
    plt.axis("off")

    # Median filter on S&P
    plt.subplot(2, 4, 7)
    plt.imshow(sp_median_filtered)
    plt.title("Median Filter 9x9")
    plt.axis("off")

    # Difference / observation
    plt.subplot(2, 4, 8)
    plt.imshow(sp_median_filtered)
    plt.title("Best Result: Median Filter")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # ========================================================
    # OBSERVATION / ANALYSIS
    # ========================================================

    print("\n" + "=" * 70)
    print("OBSERVATION AND ANALYSIS")
    print("=" * 70)

    print("""
1. Gaussian Noise:
   The 9x9 Gaussian filter reduces Gaussian noise effectively.
   It produces a smooth image because it calculates a weighted
   average of neighboring pixels.

2. Salt and Pepper Noise:
   The 9x9 Median filter performs better than the Gaussian filter.
   It removes extreme black and white impulse pixels effectively.

3. Linear Gaussian Filter:
   Since Gaussian filtering uses weighted averaging, an extreme
   noisy pixel can influence neighboring pixels. Therefore,
   it may create blurring around sharp object boundaries.

4. Non-Linear Median Filter:
   The median filter sorts the pixel values in the 9x9 neighborhood
   and selects the middle value. Extreme salt and pepper values
   are therefore rejected.

5. Edge Preservation:
   The Median filter preserves fine geometric edges better than
   the Gaussian filter, especially for impulse noise.

6. Conclusion:
   Gaussian filtering is suitable for Gaussian/random noise,
   while Median filtering is highly effective for salt-and-pepper
   impulse noise and provides better edge preservation.
""")

    print("=" * 70)
    print("EXERCISE 3 COMPLETED SUCCESSFULLY!")
    print("=" * 70)
```
