```python
# EXPERIMENT 25
# Sharpening of Image using Gradient Masking

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
filename = next(iter(uploaded))

# Read image in grayscale
img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

# Calculate Sobel gradients
gx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
gy = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

# Calculate gradient magnitude
gradient = cv2.magnitude(gx, gy)

# Convert gradient to 8-bit
gradient = cv2.convertScaleAbs(gradient)

# Create gradient mask
gradient_mask = gradient

# Sharpen using gradient mask
sharpened = cv2.add(img, gradient_mask)

# Display
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(gradient_mask, cmap='gray')
plt.title("Gradient Mask")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(sharpened, cmap='gray')
plt.title("Gradient Mask Sharpened")
plt.axis("off")

plt.show()
```
