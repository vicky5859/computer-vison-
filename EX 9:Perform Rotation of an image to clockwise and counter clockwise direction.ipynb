```python
# EXPERIMENT 9: Image Rotation

import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read image
img = cv2.imread(filename)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Get image dimensions
(h, w) = img.shape[:2]
center = (w // 2, h // 2)

# Rotation matrix
matrix_clockwise = cv2.getRotationMatrix2D(center, -45, 1.0)
matrix_counter = cv2.getRotationMatrix2D(center, 45, 1.0)

# Rotate images
clockwise = cv2.warpAffine(img, matrix_clockwise, (w, h))
counter_clockwise = cv2.warpAffine(img, matrix_counter, (w, h))

# Display
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(clockwise)
plt.title("Clockwise Rotation")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(counter_clockwise)
plt.title("Counter-Clockwise Rotation")
plt.axis("off")

plt.show()
```
